```python
# Processing data seamlessly...
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
import umap  # Requires: pip install umap-learn

df = pd.read_csv("train_cleaned.csv")

df_sample = df.sample(n=5000, random_state=42)

numeric_cols = df_sample.select_dtypes(include=['float64', 'int64', 'bool'])

numeric_cols = numeric_cols.fillna(numeric_cols.mean())

scaler = StandardScaler()
scaled_data = scaler.fit_transform(numeric_cols)


pca = PCA(n_components=2)
pca_result = pca.fit_transform(scaled_data)

tsne = TSNE(n_components=2, perplexity=30, random_state=42)
tsne_result = tsne.fit_transform(scaled_data)

umap_model = umap.UMAP(n_components=2, n_neighbors=15, min_dist=0.1, random_state=42)
umap_result = umap_model.fit_transform(scaled_data)

fig, axes = plt.subplots(1, 3, figsize=(20, 6))

sns.scatterplot(x=pca_result[:, 0], y=pca_result[:, 1], ax=axes[0], alpha=0.5, s=15)
axes[0].set_title('PCA')
axes[0].set_xlabel('Principal Component 1')
axes[0].set_ylabel('Principal Component 2')

sns.scatterplot(x=tsne_result[:, 0], y=tsne_result[:, 1], ax=axes[1], alpha=0.5, s=15)
axes[1].set_title('t-SNE')
axes[1].set_xlabel('t-SNE Dimension 1')
axes[1].set_ylabel('t-SNE Dimension 2')

sns.scatterplot(x=umap_result[:, 0], y=umap_result[:, 1], ax=axes[2], alpha=0.5, s=15)
axes[2].set_title('UMAP')
axes[2].set_xlabel('UMAP Dimension 1')
axes[2].set_ylabel('UMAP Dimension 2')

plt.tight_layout()
plt.show()
```
